```python
def detect_outliers(df, column):
    
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]
    
    return outliers
```

**Automatically Find Strong Correlations**

```python
correlation = cleaned_df[numerical_columns].corr()
correlation
```

```python
corr_pairs = correlation.unstack()

corr_pairs
```

```python
# Remove self-correlations
corr_pairs = corr_pairs[
    corr_pairs.index.get_level_values(0) !=
    corr_pairs.index.get_level_values(1)
]

corr_pairs
```

```python
corr_pairs = corr_pairs[
    corr_pairs.index.get_level_values(0) <
    corr_pairs.index.get_level_values(1)
]

corr_pairs
```

```python
# Sort the correlations
strongest_correlations = corr_pairs.sort_values(
    ascending=False
)

strongest_correlations
```

```python
strongest_pair = strongest_correlations.index[0]
strongest_value = strongest_correlations.iloc[0]

print("Strongest correlation:")
print(strongest_pair[0], "↔", strongest_pair[1])
print("Correlation:", round(strongest_value, 2))
```

```python
def classify_correlation(value):
    
    value = abs(value)
    
    if value < 0.30:
        return "Weak"
    
    elif value < 0.70:
        return "Moderate"
    
    else:
        return "Strong"
```

```python
print(classify_correlation(0.20))
print(classify_correlation(0.55))
print(classify_correlation(0.98))
print(classify_correlation(-0.85))
```

```python
correlation_report = strongest_correlations.reset_index()

correlation_report.columns = [
    "Feature_1",
    "Feature_2",
    "Correlation"
]

correlation_report["Strength"] = (
    correlation_report["Correlation"]
    .apply(classify_correlation)
)

correlation_report
```

```python
strongest = correlation_report.iloc[0]

print(
    f"{strongest['Feature_1']} and "
    f"{strongest['Feature_2']} have a "
    f"{strongest['Strength'].lower()} positive correlation "
    f"of {strongest['Correlation']:.2f}."
)
```

```python
def detect_outliers(df):

    numerical_columns = [
        "Age",
        "Salary",
        "Experience",
        "Purchase_Amount"
    ]

    outlier_report = {}

    for column in numerical_columns:

        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)

        IQR = Q3 - Q1

        lower_limit = Q1 - 1.5 * IQR
        upper_limit = Q3 + 1.5 * IQR

        outliers = df[
            (df[column] < lower_limit) |
            (df[column] > upper_limit)
        ]

        outlier_report[column] = len(outliers)

        print("=" * 40)
        print("Column:", column)
        print("Number of outliers:", len(outliers))
        print("=" * 40)

        if len(outliers) > 0:
            print(outliers[[column]])

        print()

    return outlier_report
```

```python
def get_numerical_columns(df):

    exclude_columns = [
        "Join_Year",
        "Join_Month",
        "Join_Day"
    ]

    numerical_columns = [
        column
        for column in df.select_dtypes(include="number").columns
        if column not in exclude_columns
    ]

    return numerical_columns
```

```python
def get_chart_categorical_columns(df, max_categories=10):

    categorical_columns = []

    for column in df.select_dtypes(include="object").columns:

        unique_count = df[column].nunique()

        if unique_count <= max_categories:
            categorical_columns.append(column)

    return categorical_columns
```

```python
def generate_insights(df):
    
    insights = []

    # 1. Dataset size
    rows, columns = df.shape
    
    insights.append(
        f"📊 The dataset contains {rows:,} rows and {columns} columns."
    )

    # 2. Numerical columns
    numerical_cols = df.select_dtypes(include="number").columns.tolist()

    # 3. Categorical columns
    categorical_cols = df.select_dtypes(
        include=["object", "category"]
    ).columns.tolist()

    insights.append(
        f"🔢 There are {len(numerical_cols)} numerical columns "
        f"and {len(categorical_cols)} categorical columns."
    )

    # 4. Missing values
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    if len(missing) > 0:
        for column, count in missing.items():
            percentage = (count / rows) * 100
            
            insights.append(
                f"⚠️ '{column}' contains {count:,} missing values "
                f"({percentage:.2f}% of the dataset)."
            )
    else:
        insights.append("✅ No missing values were detected.")

    # 5. Duplicate rows
    duplicates = df.duplicated().sum()

    if duplicates > 0:
        insights.append(
            f"⚠️ The dataset contains {duplicates:,} duplicate rows."
        )
    else:
        insights.append("✅ No duplicate rows were detected.")

    # 6. Numerical column insights
    for column in numerical_cols:
        
        # Skewness
        skewness = df[column].skew()

        if abs(skewness) > 1:
            if skewness > 0:
                insights.append(
                    f"📈 '{column}' is highly right-skewed."
                )
            else:
                insights.append(
                    f"📉 '{column}' is highly left-skewed."
                )

        # Outliers using IQR
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers = df[
            (df[column] < lower_bound) |
            (df[column] > upper_bound)
        ].shape[0]

        if outliers > 0:
            insights.append(
                f"⚠️ '{column}' may contain {outliers:,} potential outliers."
            )

    return insights
```

```python
insights = generate_insights(df)

for insight in insights:
    print(insight)
```

```python
def correlation_insights(df):
    
    insights = []

    numerical_cols = df.select_dtypes(include="number").columns

    # Need at least 2 numerical columns
    if len(numerical_cols) < 2:
        return ["ℹ️ Not enough numerical columns to analyze correlations."]

    correlation_matrix = df[numerical_cols].corr()

    # Check every pair of numerical columns
    for i in range(len(numerical_cols)):
        for j in range(i + 1, len(numerical_cols)):

            col1 = numerical_cols[i]
            col2 = numerical_cols[j]

            correlation = correlation_matrix.loc[col1, col2]

            if correlation >= 0.7:
                insights.append(
                    f"🔗 '{col1}' and '{col2}' have a strong positive correlation "
                    f"({correlation:.2f})."
                )

            elif correlation <= -0.7:
                insights.append(
                    f"🔗 '{col1}' and '{col2}' have a strong negative correlation "
                    f"({correlation:.2f})."
                )

    if len(insights) == 0:
        insights.append(
            "ℹ️ No strong correlations were detected between numerical columns."
        )

    return insights
```

```python
correlation_results = correlation_insights(df)

for insight in correlation_results:
    print(insight)
```

```python
def generate_summary(df):

    summary = []

    numerical_cols = df.select_dtypes(include="number").columns.tolist()

    if len(numerical_cols) == 0:
        return ["ℹ️ No numerical columns available for statistical summary."]

    for column in numerical_cols:

        min_value = df[column].min()
        max_value = df[column].max()
        mean_value = df[column].mean()
        median_value = df[column].median()

        summary.append(
            f"📊 {column}: "
            f"Min = {min_value:.2f}, "
            f"Max = {max_value:.2f}, "
            f"Mean = {mean_value:.2f}, "
            f"Median = {median_value:.2f}"
        )

    return summary
```

```python
summary_results = generate_summary(df)

for summary in summary_results:
    print(summary)
```

```python
def recommend_visualizations(df):

    recommendations = []

    numerical_cols = df.select_dtypes(include="number").columns.tolist()
    categorical_cols = df.select_dtypes(
        include=["object", "category"]
    ).columns.tolist()

    # Numerical columns
    if len(numerical_cols) >= 1:
        recommendations.append(
            "📊 Histogram: Useful for understanding the distribution of numerical columns."
        )

        recommendations.append(
            "📦 Box Plot: Useful for detecting outliers in numerical columns."
        )

    # Multiple numerical columns
    if len(numerical_cols) >= 2:
        recommendations.append(
            "🔗 Correlation Heatmap: Useful for finding relationships between numerical columns."
        )

        recommendations.append(
            "🔵 Scatter Plot: Useful for studying relationships between two numerical columns."
        )

    # Categorical columns
    if len(categorical_cols) >= 1:
        recommendations.append(
            "📊 Bar Chart: Useful for comparing categories."
        )

    # Both numerical and categorical
    if len(numerical_cols) >= 1 and len(categorical_cols) >= 1:
        recommendations.append(
            "📊 Grouped Bar Chart: Useful for comparing numerical values across categories."
        )

    if len(recommendations) == 0:
        recommendations.append(
            "ℹ️ Not enough suitable columns to recommend visualizations."
        )

    return recommendations
```

```python
def cleaning_recommendations(df):

    recommendations = []

    # -----------------------------
    # 1. Missing values
    # -----------------------------
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    if len(missing) > 0:

        for column, count in missing.items():

            percentage = (count / len(df)) * 100

            if percentage < 5:
                recommendations.append(
                    f"🧹 '{column}': {count:,} missing values "
                    f"({percentage:.2f}%). Consider filling them using "
                    f"the median/mode."
                )

            elif percentage < 30:
                recommendations.append(
                    f"⚠️ '{column}': {count:,} missing values "
                    f"({percentage:.2f}%). Consider using an appropriate "
                    f"imputation method."
                )

            else:
                recommendations.append(
                    f"🚨 '{column}': {count:,} missing values "
                    f"({percentage:.2f}%). Consider dropping the column "
                    f"if the missing data cannot be recovered."
                )

    # -----------------------------
    # 2. Duplicate rows
    # -----------------------------
    duplicates = df.duplicated().sum()

    if duplicates > 0:

        recommendations.append(
            f"🧹 Found {duplicates:,} duplicate rows. "
            f"Consider removing duplicate records."
        )

    # -----------------------------
    # 3. Outliers
    # -----------------------------
    numerical_cols = df.select_dtypes(include="number").columns.tolist()

    for column in numerical_cols:

        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)

        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers = df[
            (df[column] < lower_bound) |
            (df[column] > upper_bound)
        ].shape[0]

        if outliers > 0:

            recommendations.append(
                f"📦 '{column}': {outliers:,} potential outliers detected. "
                f"Consider investigating, capping, or treating extreme values."
            )

    # -----------------------------
    # 4. Constant columns
    # -----------------------------
    for column in df.columns:

        if df[column].nunique(dropna=False) <= 1:

            recommendations.append(
                f"🗑️ '{column}' contains only one unique value. "
                f"Consider removing this column because it provides "
                f"little or no information."
            )

    # -----------------------------
    # 5. No problems found
    # -----------------------------
    if len(recommendations) == 0:

        recommendations.append(
            "✅ No major cleaning problems were detected."
        )

    return recommendations
```

```python
cleaning_results = cleaning_recommendations(df)

for recommendation in cleaning_results:
    print(recommendation)
```
